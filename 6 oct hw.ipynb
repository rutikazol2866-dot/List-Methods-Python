```python
attendance = ("Raj","Jay","Payal","priya","Pavan","Kiran")
print(attendance)
if "priya" in attendance:
    print("priya is present")
else:
    print("priya is absent")   


for name in attendance:
    if name == "priya":
        print("priya is present")    


flag = False
for name in attendance:
    if name == "priya":
        flag =True


if flag:
    print("priya is present")
else:
    print("priya is not present")             

```
